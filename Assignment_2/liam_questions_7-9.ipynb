# Answering Questions 7-9:

```r
# Load necessary libraries
library(xts)
library(quantmod)
library(PerformanceAnalytics)
library(dplyr)
library(ggplot2)
```

```r
# Load the data
sp500_data_raw <- read.csv("sp500_2023_2024.csv")

# Remove rows with NA value in prcod
sp500_noNA <- sp500_data_raw %>%
    filter(!is.na(prcod))

# Convert datadate to Date format
sp500_noNA$datadate <- as.Date(sp500_noNA$datadate, format = "%d/%m/%Y")
sp500_data <- sp500_noNA

head(sp500_data)
```

## Question 7:
### Which ticker has the largest positive mean return?

```r
# Add daily return column next to prccd

sp500_data <- sp500_data %>%
    arrange(tic, datadate) %>%
    group_by(tic) %>%
    mutate(daily_return = (prccd / lag(prccd)) - 1) %>%
    ungroup() %>%
    relocate(daily_return, .after = prccd)

#head(sp500_data)

# Find the mean daily return for each stock

sp500_mean_returns <- sp500_data %>%
    group_by(tic) %>%
    summarize(mean_daily_return = mean(daily_return, na.rm = TRUE)) %>%
    arrange(desc(mean_daily_return))

head(sp500_mean_returns, 10)
```

### Findings:
PLTR has the highest mean daily return at 0.58%.

## Question 8:
### Which company has the largest positive mean return?

```r
# Add daily return column next to prccd

sp500_data <- sp500_data %>%
    arrange(conm, datadate) %>%
    group_by(conm) %>%
    mutate(daily_return = (prccd / lag(prccd)) - 1) %>%
    ungroup() %>%
    relocate(daily_return, .after = prccd)

#head(sp500_data)

# Find the mean daily return for each stock

sp500_mean_returns <- sp500_data %>%
    group_by(conm) %>%
    summarize(mean_daily_return = mean(daily_return, na.rm = TRUE)) %>%
    arrange(desc(mean_daily_return))

head(sp500_mean_returns, 10)
```

Palantir Technologies Inc has the highest mean daily return among the S&P 500 stocks in the dataset

## Question 9:
### Which industry is represented by the most companies?

```r
sic_companies <- sp500_data %>%
    group_by(sic) %>%
    summarize(company_count = n_distinct(conm)) %>%
    arrange(desc(company_count)) %>%
    ungroup()

head(sic_companies, 10)
```

### Findings:
The industry with SIC code 6798 (Real Estate Investment Trusts) is represented by the most companies, with a total of 28 companies in the S&P 500 between 2023 to 2024.
